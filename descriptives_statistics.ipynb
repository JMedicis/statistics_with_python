```python
norm_data = pd.DataFrame(np.random.normal(size=(1000)))
```

```python
norm_data.plot(kind='density'
               , figsize=(5,5)
)


plt.vlines(norm_data.mean()
           , ymin=0
           , ymax=0.4
           , linewidth=5
           )

plt.vlines(norm_data.median()
           , ymin=0
           , ymax=0.4
           , color='red'
           , linewidth=2
)
```

```python
skewed_data = pd.DataFrame(np.random.exponential(size=(100000)))
```

```python
skewed_data.plot(kind='density',
                figsize=(5,5),
                xlim=(-1, 5)

                )

plt.vlines(skewed_data.mean(),
            ymin=0,
            ymax=0.8,
            linewidth=5
            )

plt.vlines(skewed_data.median(),
            ymin=0,
            ymax=0.8,
            color='red',
            linewidth=2
            )
```

```python
norm_data = np.random.normal(size=50)

outliers = np.random.normal(15, size=3)

combined_data = pd.DataFrame(np.concatenate((norm_data, outliers), axis=0))
```

```python
combined_data.plot(kind='density',
                   figsize=(10,10),
                   xlim=(-5,20))  

plt.vlines(combined_data.mean(),
            ymin=0,
            ymax=0.2,
            linewidth=5,
            color='black')
            

plt.vlines(combined_data.median(),
           ymin=0,
           ymax=0.2,
           color='red',
           linewidth=2)


```

Measures of Spread

```python
max(mtcars['mpg']) - min(mtcars['mpg'])
```

```python
five_num = mtcars['mpg'].quantile([0, 0.25, 0.5, 0.75, 1])

list(five_num)
```

```python
mtcars["mpg"].quantile(0.25)
```

```python
mtcars['mpg'].describe()
```

```python
# INTERQUARTILE RANGE (IQR)

iqr = mtcars['mpg'].quantile(0.75) - mtcars['mpg'].quantile(0.25)

iqr
```

```python
mtcars.boxplot(column='mpg', 
               return_type='axes',
               figsize=(5,5))

plt.text(1.1,mtcars['mpg'].quantile(0.75), 'Q3')
plt.text(1.1, mtcars['mpg'].median(), 'Median')
plt.text(1.1, mtcars['mpg'].quantile(0.25), 'Q1')
plt.text(1.1, mtcars['mpg'].min(), 'Min')
plt.text(x=1.1, y=mtcars['mpg'].max(), s='Max')
```

```python
# Median Absolute Deviation (MAD)

abs_mediam_devs = abs(mtcars['mpg'] - mtcars['mpg'].median())


abs_mediam_devs.median() * 1.4826

# Note: The MAD is often multiplied by a scaling factor of 1.4826.
```

Skewness and Kurtosis


```python
mtcars['mpg'].skew()
```

```python
mtcars['mpg'].kurt()
```

```python
norm_data = np.random.normal(size=100000)
skewed_data = np.concatenate((np.random.normal(size=35000)+2, 
                             np.random.exponential(size=65000)), 
                             axis=0)
uniform_data = np.random.uniform(0,2, size=100000)
peaked_data = np.concatenate((np.random.exponential(size=50000),
                             np.random.exponential(size=50000)*(-1)),
                             axis=0)

data_df = pd.DataFrame({"norm":norm_data,
                       "skewed":skewed_data,
                       "uniform":uniform_data,
                       "peaked":peaked_data})
```

```python
data_df
```

```python
data_df.plot(kind='density',
             figsize=(10,5),
             xlim=(-5,5))
```

```python
data_df.skew()
```

```python
data_df.kurt()
```
